```python
import numpy as np                       # fast maths on lists of numbers
import matplotlib.pyplot as plt           # drawing the one chart our one-pager carries
```

## Step 1 — write down the model's results

We are *not* training a model today. Imagine a churn model has already been built
and validated, and it handed us a few numbers. "Churn" just means a customer
leaving or cancelling. We type the numbers in directly, exactly as a colleague
might email them to us, so the whole notebook is about the message, not the
modelling.

```python
# Set a seed so any random-looking numbers are the same every time we run this.
np.random.seed(0)

# --- The results our (already-built) churn model produced ---
total_customers          = 20000     # customers in our base
monthly_churn_rate       = 0.041     # 4.1% leave each month right now
customers_we_can_flag    = 500       # the model flags the 500 likeliest leavers
fraction_we_can_save     = 0.30      # past pilots saved about 30% of those flagged
revenue_per_customer_yr  = 8000      # each retained customer is worth Rs 8000/year

print("Customers in the base        :", total_customers)
print("Current monthly churn rate   :", monthly_churn_rate)
print("Customers the model flags    :", customers_we_can_flag)
print("Fraction we expect to retain :", fraction_we_can_save)
print("Yearly value per customer    : Rs", revenue_per_customer_yr)
```

## Step 2 — turn a metric into a rupee figure

A churn rate or a probability means nothing to a manager. The skill is to translate
it into money. Here we work out, one plain step at a time, how much revenue the
model could protect in a year.

```python
# How many of the flagged customers do we expect to actually keep?
customers_retained = customers_we_can_flag * fraction_we_can_save

# Each retained customer is worth this much revenue per year.
revenue_protected_per_year = customers_retained * revenue_per_customer_yr

# Show the chain of reasoning, one line at a time.
print("Customers flagged          :", customers_we_can_flag)
print("Expected to be retained    :", customers_retained)
print("Revenue protected per year : Rs", revenue_protected_per_year)

# The same number in crore, which is how a senior leader will hear it.
# 1 crore = 10,000,000.
revenue_protected_crore = revenue_protected_per_year / 1e7
print("Revenue protected per year : Rs", round(revenue_protected_crore, 2), "crore")
```

## Step 3 — the one sentence that matters

Before drawing anything, decide the single sentence the reader must remember.
Everything else on the page will serve this sentence. This is the "answer" we will
lead with.

```python
# This is the "bottom line" we will put first.
headline = ("Targeting the 500 likeliest leavers protects about Rs "
            + str(round(revenue_protected_crore, 2)) + " crore a year.")

print(headline)
```

## Step 4 — one clean summary chart

A one-pager carries exactly ONE chart: the one that makes the argument. Here that
is the money. What do we lose if we do nothing, versus what do we protect if we
act? We keep it plain: two bars, clear labels, no decoration.

```python
# The two outcomes we want to compare, in rupees per year.
do_nothing = 0                                  # we protect nothing
act_now    = revenue_protected_per_year         # we protect this much

labels  = ["Do nothing", "Run the pilot"]
values  = [do_nothing, act_now]
colors  = ["#B0B0B0", "#2E75B6"]                # grey vs blue: acting is the point

plt.figure(figsize=(6, 5))
bars = plt.bar(labels, values, color=colors, width=0.6)

# Write the rupee value on top of each bar so the reader needs no axis-reading.
for one_bar, value in zip(bars, values):
    plt.text(one_bar.get_x() + one_bar.get_width() / 2,
             value + max(values) * 0.02,
             "Rs " + format(int(value), ","),
             ha="center", va="bottom", fontsize=11)

plt.ylabel("revenue protected per year (Rs)")
plt.title("Acting on the churn model protects revenue")
plt.ylim(0, max(values) * 1.18)                 # headroom for the labels
plt.show()
```

## Step 5 — the executive one-pager (recommendation first / BLUF)

BLUF means **B**ottom **L**ine **U**p **F**ront: lead with the recommendation, then
justify it. A non-technical reader should get the whole decision from the first
sentence. We build the one-pager as a list of lines and print it, so you can read
it like the email a manager would receive. Notice the order: recommendation first,
evidence second, caveat and next step last.

```python
# Each entry is one line of the one-pager. The ORDER is the whole point.
one_pager_lines = [
    "RECOMMENDATION",
    "  Approve a 3-month retention pilot targeting the 500 likeliest leavers.",
    "  Expected benefit: about Rs " + str(round(revenue_protected_crore, 2)) + " crore of revenue protected per year.",
    "",
    "WHY (evidence)",
    "  - Churn is running at " + str(round(monthly_churn_rate * 100, 1)) + "% a month across " + format(total_customers, ",") + " customers.",
    "  - The model can rank customers by risk and flag the top " + str(customers_we_can_flag) + ".",
    "  - Similar past pilots retained about " + str(int(fraction_we_can_save * 100)) + "% of those contacted.",
    "",
    "RISK / CAVEAT",
    "  - The 30% save-rate is an estimate from earlier pilots; we will measure it for real.",
    "",
    "NEXT STEP",
    "  - Owner: Retention team. Start: next month. Review the results after 3 months.",
]

for line in one_pager_lines:
    print(line)
```

## Step 6 — render the same one-pager as tidy Markdown

Jupyter can display Markdown from code using `IPython.display`. This is the version
you would paste into a document or an email. Same words, prettier.

```python
from IPython.display import Markdown, display   # lets a code cell show formatted text

# A triple-quoted string lets us write several lines, including Markdown headings.
one_pager_markdown = """
### Cut churn: approve the retention pilot

**Recommendation.** Approve a 3-month retention pilot targeting the 500 likeliest
leavers. Expected benefit: about **Rs {crore} crore** of revenue protected per year.

**Why.**
- Churn is running at {rate}% a month across {base} customers.
- The model flags the top {flagged} customers most likely to leave.
- Past pilots retained about {save}% of those contacted.

**Risk.** The 30% save-rate is an estimate; the pilot will measure the real figure.

**Next step.** Retention team to start next month; review after 3 months.
""".format(
    crore=round(revenue_protected_crore, 2),
    rate=round(monthly_churn_rate * 100, 1),
    base=format(total_customers, ","),
    flagged=customers_we_can_flag,
    save=int(fraction_we_can_save * 100),
)

display(Markdown(one_pager_markdown))
```

### Stretch (optional) — turn the single number into an honest range

Skip this if you are new to code. The Rs 1.2 crore figure came from one guess: that
we save 30% of the people we contact. In real life that save-rate is uncertain. A
more honest headline gives a range instead of a single confident number, using a
pessimistic and an optimistic save-rate. This is the same "show a range, not a
false-precision point" idea you will meet again with the fan chart in the next
notebook.

```python
# A cautious and a hopeful version of the save-rate.
low_save_rate  = 0.20
high_save_rate = 0.40

# The revenue protected under each, in crore.
low_revenue_crore  = customers_we_can_flag * low_save_rate  * revenue_per_customer_yr / 1e7
high_revenue_crore = customers_we_can_flag * high_save_rate * revenue_per_customer_yr / 1e7

print("Pessimistic (20% saved): Rs", round(low_revenue_crore, 2), "crore")
print("Optimistic  (40% saved): Rs", round(high_revenue_crore, 2), "crore")
print()
print("Honest headline: 'protects roughly Rs", round(low_revenue_crore, 2),
      "to", round(high_revenue_crore, 2), "crore a year'.")
```
