```python
import numpy as np                       # fast maths on lists of numbers
import matplotlib.pyplot as plt           # drawing the one chart in our one-page summary
```

## Step 1: write down the model's results

We are *not* training a model today. Imagine a churn model is already built and
checked, and it gave us a few numbers. "Churn" means a customer leaving or
cancelling. We type the numbers in directly, as a colleague might send them by
email. So this notebook is about the message, not about building the model.

```python
# Set a seed so any random-looking numbers are the same every time we run this.
np.random.seed(0)

# --- The results our (already-built) churn model produced ---
total_customers          = 20000     # customers in our base
monthly_churn_rate       = 0.041     # 4.1% leave each month right now
customers_we_can_flag    = 500       # the model flags the 500 customers most likely to leave
fraction_we_can_save     = 0.30      # past pilots kept about 30% of those flagged
revenue_per_customer_yr  = 8000      # each customer we keep is worth Rs 8000 a year

print("Customers in the base        :", total_customers)
print("Current monthly churn rate   :", monthly_churn_rate)
print("Customers the model flags    :", customers_we_can_flag)
print("Share we expect to keep      :", fraction_we_can_save)
print("Yearly value per customer    : Rs", revenue_per_customer_yr)
```

## Step 2: turn a model number into a rupee amount

A churn rate or a probability means nothing to a manager who is not technical. The
skill is to translate it into money. Here we work out, one step at a time, how
much revenue the model could protect in a year.

```python
# How many of the flagged customers do we expect to keep?
customers_retained = customers_we_can_flag * fraction_we_can_save

# Each customer we keep is worth this much revenue per year.
revenue_protected_per_year = customers_retained * revenue_per_customer_yr

# Show the working, one line at a time.
print("Customers flagged          :", customers_we_can_flag)
print("Expected to be kept        :", customers_retained)
print("Revenue protected per year : Rs", revenue_protected_per_year)

# The same number in crore. 1 crore = 100 lakh = 10,000,000.
# 0.12 crore is the same as Rs 12 lakh.
revenue_protected_crore = revenue_protected_per_year / 1e7
print("Revenue protected per year : Rs", round(revenue_protected_crore, 2), "crore")
```

## Step 3: the one sentence that matters

Before drawing anything, decide the one sentence the reader must remember.
Everything else on the page supports this sentence. This is the answer we will
put first.

```python
# This is the "bottom line", the answer we will put first.
headline = ("Calling the 500 customers most likely to leave protects about Rs "
            + str(round(revenue_protected_crore * 100)) + " lakh a year.")

print(headline)
```

## Step 4: one clean summary chart

A one-page summary has only ONE chart: the chart that supports the recommendation.
Here that chart is about money. How much do we protect if we do nothing, and how
much if we act? We keep it plain: two bars, clear labels, no decoration.

```python
# The two outcomes we want to compare, in rupees per year.
do_nothing = 0                                  # we protect nothing
act_now    = revenue_protected_per_year         # we protect this much

labels  = ["Do nothing", "Run the pilot"]
values  = [do_nothing, act_now]
colors  = ["#B0B0B0", "#2E75B6"]                # grey vs blue: blue is the action we recommend

plt.figure(figsize=(6, 5))
bars = plt.bar(labels, values, color=colors, width=0.6)

# Write the rupee value on top of each bar so the reader needs no axis-reading.
for one_bar, value in zip(bars, values):
    plt.text(one_bar.get_x() + one_bar.get_width() / 2,
             value + max(values) * 0.02,
             "Rs " + format(int(value), ","),
             ha="center", va="bottom", fontsize=11)

plt.ylabel("revenue protected per year (Rs)")
plt.title("Acting on the churn model protects revenue")
plt.ylim(0, max(values) * 1.18)                 # headroom for the labels
plt.show()
```

## Step 5: the one-page summary (answer first, BLUF)

Answer first is also called BLUF: **B**ottom **L**ine **U**p **F**ront. Give the
recommendation first, then the reasons. A manager who is not technical should get
the whole decision from the first sentence.

We build the one-page summary as a list of lines and print it. Read it like an
email a manager would receive. Look at the order: recommendation first, evidence
second, risk and next step last.

```python
# Each entry is one line of the one-page summary. The ORDER of the lines matters most.
one_pager_lines = [
    "RECOMMENDATION",
    "  Approve a 3-month retention pilot on the 500 customers most likely to leave.",
    "  Expected benefit: about Rs " + str(round(revenue_protected_crore * 100)) + " lakh of revenue protected per year.",
    "",
    "WHY (evidence)",
    "  - Churn is running at " + str(round(monthly_churn_rate * 100, 1)) + "% a month across " + format(total_customers, ",") + " customers.",
    "  - The model can rank customers by risk and flag the top " + str(customers_we_can_flag) + ".",
    "  - Similar past pilots kept about " + str(int(fraction_we_can_save * 100)) + "% of those contacted.",
    "",
    "RISK",
    "  - The 30% is an estimate from earlier pilots; the pilot will measure the real figure.",
    "",
    "NEXT STEP",
    "  - Owner: Retention team. Start: next month. Review the results after 3 months.",
]

for line in one_pager_lines:
    print(line)
```

## Step 6: show the same one-page summary as neat Markdown

Jupyter can show formatted Markdown text from code, using `IPython.display`. This
is the version you would paste into a document or an email. The words are the
same; only the look is neater.

```python
from IPython.display import Markdown, display   # lets a code cell show formatted text

# A triple-quoted string lets us write several lines, including Markdown headings.
one_pager_markdown = """
### Keep about Rs {lakh} lakh a year: approve the retention pilot

**Recommendation.** Approve a 3-month retention pilot on the 500 customers most likely
to leave. Expected benefit: about **Rs {lakh} lakh** of revenue protected per year.

**Why.**
- Churn is running at {rate}% a month across {base} customers.
- The model flags the top {flagged} customers most likely to leave.
- Past pilots kept about {save}% of those contacted.

**Risk.** The 30% is an estimate; the pilot will measure the real figure.

**Next step.** Retention team to start next month; review after 3 months.
""".format(
    lakh=round(revenue_protected_crore * 100),
    rate=round(monthly_churn_rate * 100, 1),
    base=format(total_customers, ","),
    flagged=customers_we_can_flag,
    save=int(fraction_we_can_save * 100),
)

display(Markdown(one_pager_markdown))
```

### Stretch (optional): turn the single number into an honest range

Skip this if you are new to code. The figure of about Rs 12 lakh a year (0.12
crore) came from one guess: that we keep 30% of the customers we contact. In real
life we do not know this share exactly.

A more honest headline gives a likely range, not one confident number. We use a
low guess (20% kept) and a high guess (40% kept). This gives about Rs 8 lakh to
Rs 16 lakh a year (0.08 to 0.16 crore). You will see the same idea again in the
next notebook, as a fan chart: show a range, not one number that looks more exact
than it is.

```python
# A low guess and a high guess for the share of flagged customers we keep.
low_save_rate  = 0.20
high_save_rate = 0.40

# The revenue protected for each guess, in crore.
low_revenue_crore  = customers_we_can_flag * low_save_rate  * revenue_per_customer_yr / 1e7
high_revenue_crore = customers_we_can_flag * high_save_rate * revenue_per_customer_yr / 1e7

print("Low guess  (20% kept): Rs", round(low_revenue_crore * 100), "lakh")
print("High guess (40% kept): Rs", round(high_revenue_crore * 100), "lakh")
print()
print("Honest headline: 'protects about Rs", round(low_revenue_crore * 100),
      "to", round(high_revenue_crore * 100), "lakh a year'.")
```
