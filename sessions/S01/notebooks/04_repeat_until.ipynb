# S01 · Repeat until you get there (the `while` loop)

A `for` loop repeats a fixed number of times. Sometimes you do not know how many steps
you need — you just repeat *until* something happens. That is a `while` loop. We use it
on a real money question: how many years until savings double?

## Step 1 — the question

You put Rs 10,000 in a savings pot that grows 8% each year. How many years until it
reaches Rs 20,000? We do not know the answer up front, so we keep growing it *until* it
passes the target.

```python
balance = 10000
yearly_growth = 0.08
target = 20000

years = 0
while balance < target:
    balance = balance * (1 + yearly_growth)
    years = years + 1

print("years to reach Rs", target, ":", years)
print("balance then      : Rs", round(balance))
```

## Step 2 — watch it grow

Run the same loop again, but this time record the balance each year so we can see it.

```python
balance = 10000
history = [balance]
while balance < target:
    balance = balance * (1 + yearly_growth)
    history.append(balance)

print("balance each year:", [round(b) for b in history])
```

```python
plt.figure(figsize=(7, 4))
plt.plot(range(len(history)), history, marker="o", color="#2E75B6")
plt.axhline(target, color="#C0392B", linestyle="--", label="target Rs 20,000")
plt.xlabel("year"); plt.ylabel("balance (Rs)")
plt.title("Savings growing at 8% a year")
plt.legend(); plt.show()
```

## Step 3 — a reusable recipe (a function)

Wrap the idea so we can ask it for any starting amount, rate and target.

```python
def years_to_reach(start, rate, target):
    balance = start
    years = 0
    while balance < target:
        balance = balance * (1 + rate)
        years = years + 1
    return years

print("Rs 10,000 -> 20,000 at 8%:", years_to_reach(10000, 0.08, 20000), "years")
print("Rs 10,000 -> 20,000 at 12%:", years_to_reach(10000, 0.12, 20000), "years")
```

## Step 4 — a second everyday 'repeat until': stock running low

A shelf starts with 100 units and sells 15 a day. How many days until it drops below the
reorder level of 20? Again we repeat *until* a condition is met.

```python
stock = 100
sold_per_day = 15
reorder_level = 20

day = 0
while stock >= reorder_level:
    stock = stock - sold_per_day
    day = day + 1

print("reorder on day", day, "when stock falls to", stock)
```

### Stretch (optional) — zooming in on the square root of 2

Skip unless curious. A `while` loop can also 'zoom in' on a number: start with a guess
and keep improving it until it barely changes. This finds the square root of 2.

```python
guess = 1.0
while abs(guess * guess - 2) > 1e-12:
    guess = (guess + 2 / guess) / 2   # the classic improve-the-guess step
print("square root of 2 is about:", guess)
```
