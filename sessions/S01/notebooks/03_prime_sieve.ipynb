## Step 1 — what a prime is, and the idea of the sieve

A prime number is a whole number above 1 whose only exact divisors are 1 and
itself: 2, 3, 5, 7, 11, 13, and so on. Primes are the building blocks of every
other whole number, and they quietly protect you every day: the encryption behind
UPI payments and bank logins is built on the difficulty of splitting a huge number
back into its prime factors.

The sieve idea is lovely. Start by assuming *every* number is prime. Then walk
through 2, 3, 4, and so on, and whenever a number is still standing, cross out all
of its multiples, because a multiple has an extra divisor and so cannot be prime.
Whatever is never crossed out is prime.

We track "still standing?" with a list of `True`/`False` flags, one per number.

```python
# We will find all primes up to and including this limit.
limit = 50

# Assume every number from 0 to 50 is prime: a list of 51 True flags.
is_prime = [True] * (limit + 1)

# 0 and 1 are not prime, so cross them out straight away.
is_prime[0] = False
is_prime[1] = False

print("We made", len(is_prime), "flags, all starting True.")
print("First few flags:", is_prime[:8])
```

## Step 2 — cross out the multiples

Now walk through the numbers. For each one that is still standing, cross out every
multiple of it, because those multiples cannot be prime. We do this with one loop
inside another loop.

```python
# Go through each candidate from 2 up to the limit.
for number in range(2, limit + 1):
    # Only act on numbers that are still standing.
    if is_prime[number] == True:
        # Cross out every multiple of "number", starting at number + number.
        multiple = number * 2
        while multiple <= limit:
            is_prime[multiple] = False
            multiple = multiple + number

print("Done crossing out. The survivors are the primes.")
```

## Step 3 — collect and print the survivors

Finally, look at every flag. Wherever it is still `True`, that number survived the
sieve, so it is prime. We add each survivor to a list and print the list.

```python
primes = []
for number in range(0, limit + 1):
    if is_prime[number] == True:
        primes.append(number)

print("Primes up to", limit, ":")
print(primes)
print("There are", len(primes), "of them.")
```

### Stretch (optional) — collect the primes in one line

Skip this if you are new to code. If you already program, that whole collecting
loop is one comprehension. It reads as "each number in the range, kept only where
its flag is still True", which is set-builder notation in disguise.

```python
primes_short = [number for number in range(2, limit + 1) if is_prime[number]]
print(primes_short)
```

## Step 4 — a picture of which numbers survived

For fun, a simple plot: a green dot for every prime and a faint grey dot for every
non-prime, from 2 to 50. You can see the primes thin out as the numbers grow.

```python
# Sort the numbers into two lists for the plot.
prime_numbers = []
other_numbers = []
for number in range(2, limit + 1):
    if is_prime[number] == True:
        prime_numbers.append(number)
    else:
        other_numbers.append(number)

plt.figure(figsize=(9, 2))
plt.scatter(other_numbers, [0] * len(other_numbers),
            color="#CCCCCC", label="not prime")
plt.scatter(prime_numbers, [1] * len(prime_numbers),
            color="#2E9E5B", label="prime")
plt.yticks([0, 1], ["not prime", "prime"])
plt.xlabel("number")
plt.title("Primes up to 50 (the survivors of the sieve)")
plt.legend(loc="center left", bbox_to_anchor=(1.0, 0.5))
plt.tight_layout()
plt.show()
```
