## Step 1 — the idea of the sieve
To find primes up to some limit, start by assuming **every** number is prime. Then go through 2, 3, 4, … and whenever a number is still standing, cross out all of its multiples (they cannot be prime). Whatever is never crossed out is prime.

We track this with a list of True/False flags: `is_prime[n]` is `True` while the number `n` is still standing.

```python

# We will find all primes up to (and including) this limit.
limit = 50

# Start by assuming every number from 0 to 50 is prime.
# This makes a list of 51 True values: one flag for each number 0..50.
is_prime = [True] * (limit + 1)

# 0 and 1 are not prime, so cross them out straight away.
is_prime[0] = False
is_prime[1] = False

print("We made a list of", len(is_prime), "flags, all starting True.")
print("First few flags:", is_prime[:8])
```

## Step 2 — cross out the multiples
Now we walk through the numbers. For each number that is still standing (its flag is True), we cross out every multiple of it. We do this with one loop inside another loop.

```python

# Go through each possible prime, from 2 up to the limit.
for number in range(2, limit + 1):
    # Only act on numbers that are still standing.
    if is_prime[number] == True:
        # Cross out every multiple of "number", starting at number * 2.
        multiple = number * 2
        while multiple <= limit:
            is_prime[multiple] = False
            multiple = multiple + number

print("Done crossing out. The survivors are the primes.")
```

## Step 3 — collect and print the primes
Finally we look at every flag. Wherever the flag is still `True`, that number survived the sieve, so it is prime. We add it to a list and print the list.

```python

# Build a list of the numbers whose flag is still True.
primes = []
for number in range(0, limit + 1):
    if is_prime[number] == True:
        primes.append(number)

print("Primes up to", limit, ":")
print(primes)
print("There are", len(primes), "of them.")
```

## Step 4 — optional picture: which numbers survived
This is a simple plot for fun: a green dot for every prime, a faint grey dot for every non-prime, from 2 to 50. You can see the primes thin out as the numbers grow.

```python

import matplotlib.pyplot as plt

# Separate the numbers into primes and non-primes for plotting.
prime_numbers = []
other_numbers = []
for number in range(2, limit + 1):
    if is_prime[number] == True:
        prime_numbers.append(number)
    else:
        other_numbers.append(number)

plt.figure(figsize=(9, 2))
# Non-primes on the bottom row, primes on the top row.
plt.scatter(other_numbers, [0] * len(other_numbers), color="#CCCCCC", label="not prime")
plt.scatter(prime_numbers, [1] * len(prime_numbers), color="#2E9E5B", label="prime")
plt.yticks([0, 1], ["not prime", "prime"])
plt.xlabel("number")
plt.title("Primes up to 50 (survivors of the sieve)")
plt.legend(loc="center left", bbox_to_anchor=(1.0, 0.5))
plt.tight_layout()
plt.show()
```

## What you just did
You used a list of True/False flags and two loops to find every prime up to 50. The list-of-flags trick — keep a yes/no answer for each item, then update it — is a workhorse you will reuse all through the course.
