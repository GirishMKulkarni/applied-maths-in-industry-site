## Step 1 — a variable is just a name for a value

The `=` sign in code does not mean "solve for x". Read it left to right as an
instruction: *from now on, this name stands for this value.* Here we give the
name `price_per_item` to the number 249, so we can use it again later without
retyping it.

`print(...)` shows a value on the screen. Use it often so you can see what is
happening.

```python
# Give the name "price_per_item" to the value 249 (rupees).
price_per_item = 249

# Show it on the screen.
print("price per item:", price_per_item)
```

## Step 2 — a name can point at a new value

Say a customer buys 3 items. We can compute the bill and put it in its own name.
Notice the second line: `total = total + shipping`. As an equation that looks
impossible, but as an instruction it is simple. *Work out the old total plus the
shipping, then let the name `total` point at that new value.* This "keep a running
total" trick is how we add up a day's orders later.

```python
quantity = 3
shipping = 40

# Work out the bill, step by step.
total = price_per_item * quantity   # 249 * 3
total = total + shipping            # add delivery on top

print("items cost:", price_per_item * quantity)
print("total bill:", total)
```

## Step 3 — the four basic kinds of value

Every value has a *type*, meaning what kind of thing it is. The four you meet
first cover almost everything:

- `int` — a whole number, like a count of orders.
- `float` — a number with a decimal point, like a price.
- `str` — text, always inside quotes (a "string" of characters).
- `bool` — a yes/no value, either `True` or `False`.

`type(...)` asks Python "what kind of value is this?" You rarely need it, but it
is nice to see the labels once.

```python
number_of_orders = 42            # int  (a whole number)
average_bill = 517.50            # float (has a decimal point)
top_city = "Pune"                # str  (text, in quotes)
paid_online = True               # bool (True or False)

print("number_of_orders is a", type(number_of_orders))
print("average_bill     is a", type(average_bill))
print("top_city         is a", type(top_city))
print("paid_online      is a", type(paid_online))
```

## Step 4 — choosing a path with `if` / `else`

Often a program must decide what to do. We ask a yes/no question (a *condition*)
and follow one branch or the other. Here the rule is: orders above ₹1000 get free
delivery. The lines that belong to a branch are **indented** underneath it. In
Python those spaces are not decoration; they are how the computer knows which
lines are inside the `if`.

```python
order_amount = 1250

if order_amount > 1000:
    # This line runs only when the amount is above 1000.
    print("Order of", order_amount, ": free delivery")
else:
    # This line runs only when it is not.
    print("Order of", order_amount, ": delivery charged")
```

## Step 5 — repeating work with a `for` loop

Here is the real payoff for our store. We have a list of the day's order amounts,
and we want the total. A `for` loop walks through the list one item at a time and
runs the indented lines once for each. We keep a running total, exactly like in
Step 2.

```python
# One evening's orders, in rupees.
order_amounts = [249, 1250, 517, 980, 1499, 340]

# Start the total at zero, then add each order in turn.
total_sales = 0
for one_order in order_amounts:
    total_sales = total_sales + one_order

print("number of orders:", len(order_amounts))
print("total sales     :", total_sales)
```

## Step 6 — counting with a `for` loop and an `if`

Combine the two moves. Walk through the same orders, and each time the amount is
above ₹1000, add one to a counter. This is the second number the store owner
wanted: how many orders qualified for free delivery.

```python
free_delivery_count = 0
for one_order in order_amounts:
    if one_order > 1000:
        free_delivery_count = free_delivery_count + 1

print("orders above 1000:", free_delivery_count)
```

## Step 7 — a function is a reusable recipe

A function lets you name a calculation once and use it as many times as you like,
without copying and pasting. You *define* it with `def`, and the `return` line
hands the answer back. Here we write a recipe that adds 18% GST to any price.

```python
# Define the recipe once.
def price_with_gst(amount):
    tax = amount * 0.18
    return amount + tax

# Now use it on several prices, just by calling its name.
print("249 with GST :", price_with_gst(249))
print("1250 with GST:", price_with_gst(1250))
print("999 with GST :", price_with_gst(999))
```

## Step 8 — a list and a lookup table (dict)

Two containers you will use constantly. A **list** holds several values in order,
in square brackets. A **dictionary** (or *dict*) is a lookup table: you give it a
key and it hands back the matching value, written with curly braces and
`key: value` pairs. Here a dict remembers how many orders came from each city.

```python
# A list: the cities that ordered today, in the order they came in.
cities_today = ["Pune", "Mumbai", "Pune", "Nagpur", "Pune", "Mumbai"]
print("first city today:", cities_today[0])   # counting starts at 0
print("how many orders :", len(cities_today))

# A dict: a small lookup table of orders per city.
orders_per_city = {"Pune": 3, "Mumbai": 2, "Nagpur": 1}
print("orders from Pune:", orders_per_city["Pune"])
```

### Stretch (optional) — the compact way, and a note for maths readers

Skip this if you are new to code. If you already program, here is the short form.
A **comprehension** builds a new list in one line. And a **set** keeps only the
distinct values, throwing away order and duplicates, which answers "which cities
ordered at all?" instantly.

For readers who enjoy the maths link: that comprehension is exactly set-builder
notation. `[amount for amount in order_amounts if amount > 1000]` reads as
"the amounts, for each amount in the list, where the amount exceeds 1000", the
same sentence as `{ a : a ∈ orders, a > 1000 }`. And a Python set behaves like a
set in set theory: no order, no repeats.

```python
# The big orders, in one line instead of a loop.
big_orders = [amount for amount in order_amounts if amount > 1000]
print("big orders:", big_orders)

# The distinct cities, with duplicates and order dropped.
distinct_cities = set(cities_today)
print("distinct cities:", distinct_cities)
```
