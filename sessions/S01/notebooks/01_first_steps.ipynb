## Step 1 — a variable is just a name for a value
In maths you write *let n = 5*. In Python the `=` sign means the same thing: it is an instruction that says *from now on, the name `n` refers to the value 5*. It is **not** an equation to solve.

Run the next cell. `print(...)` shows a value on the screen.

```python

# Give the name "number_of_apples" to the value 5.
number_of_apples = 5

# Show it on the screen.
print("number of apples:", number_of_apples)
```

## Step 2 — change what a name points to
`number_of_apples = number_of_apples + 1` looks strange as an equation, but read it left to right as an instruction: *compute the old value plus one, then let the name point at that new value*.

```python

# Add one more apple. Read this as: new value = old value + 1.
number_of_apples = number_of_apples + 1

print("now we have:", number_of_apples)
```

## Step 3 — the four basic types
Every value in Python has a *type*. The four you meet first map directly onto things you already know:

- `int` — a whole number (your integers)
- `float` — a decimal number (a real number, approximately)
- `str` — text in quotes (a 'string' of characters)
- `bool` — a truth value, `True` or `False` (a proposition)

`type(...)` asks Python *what kind of value is this?*

```python

# One value of each type, with a clear name.
a_whole_number = 7              # int
a_decimal_number = 3.14         # float
some_text = "Nifty50"           # str
a_truth_value = True            # bool

# Ask Python the type of each one.
print("a_whole_number   is a", type(a_whole_number))
print("a_decimal_number is a", type(a_decimal_number))
print("some_text        is a", type(some_text))
print("a_truth_value    is a", type(a_truth_value))
```

## Step 4 — choosing a path with `if` / `else`
Sometimes a program must choose what to do. We test a condition (a yes/no question) and follow one branch or the other. The lines that belong to a branch are **indented** underneath it — in Python the indentation is how the computer knows what is inside the `if`.

```python

# Pick a number and decide whether it is positive.
x = -3

if x > 0:
    # This line runs only when the test x > 0 is True.
    print(x, "is positive")
else:
    # This line runs only when the test is False.
    print(x, "is not positive")
```

## Step 5 — repeating work with a `for` loop
A `for` loop is the mathematician's *for each*. `range(1, 6)` gives the numbers 1, 2, 3, 4, 5 (it stops just before 6). For each one, we run the indented body once.

```python

# Print the square of each number from 1 to 5.
for number in range(1, 6):
    square = number * number
    print("the square of", number, "is", square)
```

## Step 6 — defining your own function
A function in code is exactly a function in maths: you hand it an input, it gives back one output. Here we write `f(x) = x²`. We *define* it once with `def`, then *call* it many times. The `return` line is how the function hands its answer back to us.

```python

# Define the function f(x) = x squared.
def square_of(x):
    result = x * x
    return result

# Call it on a few inputs, just like evaluating f at several points.
print("square_of(2) =", square_of(2))
print("square_of(3) =", square_of(3))
print("square_of(10) =", square_of(10))
```

## Step 7 — a function is a mapping (domain → value)
In your maths classes a function sends each input to exactly one output. Let us show that table of values for `f(x) = x²` on the domain {0, 1, 2, 3, 4}. Same input always gives the same output — that is what makes it a function.

```python

# Show the input -> output table for f on the domain 0..4.
for input_value in range(0, 5):
    output_value = square_of(input_value)
    print("f(", input_value, ") =", output_value)
```
