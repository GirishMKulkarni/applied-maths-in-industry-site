## Step 1 — sales by city (a dictionary)

A **dictionary** maps a key to a value. Here the key is a city and the value is its
sales in rupees. You look a value up by its key.

```python
city_sales = {"Pune": 42000, "Mumbai": 88000, "Delhi": 51000, "Nagpur": 17000}
print("cities:", list(city_sales))
print("Mumbai sales:", city_sales["Mumbai"])
```

## Step 2 — the grand total (loop over the values)

```python
grand_total = 0
for city in city_sales:
    grand_total = grand_total + city_sales[city]
print("grand total across all cities: Rs", grand_total)
```

## Step 3 — which city bought the most?

Keep a 'best so far' city, and replace it whenever we find one with higher sales.

```python
top_city = None
top_value = -1
for city in city_sales:
    if city_sales[city] > top_value:
        top_value = city_sales[city]
        top_city = city
print("top city:", top_city, "with Rs", top_value)
```

## Step 4 — label each customer (a function with `if` / `elif` / `else`)

A function that turns a customer's total spend into a loyalty tier.

```python
def tier(spend):
    if spend >= 5000:
        return "gold"
    elif spend >= 2000:
        return "silver"
    else:
        return "bronze"

customers = {"Asha": 6200, "Ravi": 2500, "Neha": 800, "Imran": 5000}
for name in customers:
    print(name, "spent Rs", customers[name], "->", tier(customers[name]))
```

## Step 5 — a tidy printed report

Loop over the cities and print each, then the headline numbers.

```python
print("=== Sales report ===")
for city in city_sales:
    print(" ", city, ":", "Rs", city_sales[city])
print("Total          : Rs", grand_total)
print("Best city      :", top_city)
```

### Stretch (optional) — a classic crossing-out pattern (the prime sieve)

Skip unless curious. The same loop-and-list idea finds every prime up to 50 by crossing
out multiples, the way you might with a pencil.

```python
limit = 50
is_prime = [True] * (limit + 1)
is_prime[0] = is_prime[1] = False
for number in range(2, limit + 1):
    if is_prime[number]:
        for multiple in range(number * 2, limit + 1, number):
            is_prime[multiple] = False
primes = []
for number in range(2, limit + 1):
    if is_prime[number]:
        primes.append(number)
print("primes up to 50:", primes)
```
