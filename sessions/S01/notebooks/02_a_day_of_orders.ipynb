## Step 1 — a day's orders as a list

Each number is one order, in rupees. A **list** holds them in order, in square brackets.

```python
orders = [250, 1200, 90, 4300, 760, 1500, 320, 999, 2100, 180]   # rupees
print("number of orders:", len(orders))
print("the first order  :", orders[0])
```

## Step 2 — total sales (a running total with a `for` loop)

Start a total at 0, then walk through the orders one at a time and add each to it.

```python
total_sales = 0
for one_order in orders:
    total_sales = total_sales + one_order
print("total sales today: Rs", total_sales)
```

## Step 3 — how many orders earned free delivery (above Rs 1000)?

Same loop, but this time we *count* — we add 1 only when an order clears the bar.

```python
free_delivery_count = 0
for one_order in orders:
    if one_order > 1000:
        free_delivery_count = free_delivery_count + 1
print("orders with free delivery:", free_delivery_count)
```

## Step 4 — the biggest order of the day

Keep a 'best so far'. Start it at the first order, then replace it whenever we see a bigger one.

```python
biggest_order = orders[0]
for one_order in orders:
    if one_order > biggest_order:
        biggest_order = one_order
print("biggest order: Rs", biggest_order)
```

## Step 5 — a reusable recipe: add GST (a function)

A **function** is a named recipe: an input goes in, one answer comes back. We write it
once and reuse it. Here it adds 18% GST to any amount.

```python
def add_gst(amount):
    # add 18 percent GST to a price
    return amount * 1.18

print("Rs 1000 with GST:", add_gst(1000))
print("today's total with GST:", round(add_gst(total_sales), 2))
```

## Step 6 — the report in one line

Put the numbers we found into a single sentence, the way the owner wants it.

```python
print("Today:", len(orders), "orders,", "Rs", total_sales, "in sales,",
      free_delivery_count, "with free delivery, biggest was Rs", biggest_order, ".")
```

### Stretch (optional) — two classic number patterns

Skip this unless you are curious. The same moves (a loop, a running value) also make the
famous **Fibonacci** sequence, and **Euclid's method** for the largest number that divides
two others (the GCD). Same tools, a number-theory flavour.

```python
# Fibonacci: each number is the sum of the previous two.
a, b = 0, 1
fibonacci = [a, b]
for step in range(8):
    a, b = b, a + b
    fibonacci.append(b)
print("Fibonacci:", fibonacci)

# Euclid's GCD: keep replacing the bigger by its remainder until one hits 0.
x, y = 48, 36
while y != 0:
    x, y = y, x % y
print("largest common factor of 48 and 36:", x)
```
