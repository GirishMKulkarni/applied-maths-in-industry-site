## Step 1 — the function and its derivative
We want the value of x where `f(x) = x² − 2` equals 0. That value is exactly √2. Newton–Raphson needs the function and its derivative `f'(x) = 2x`, so we write both as small Python functions.

```python

# The function whose root we want: f(x) = x^2 - 2.
def f(x):
    return x * x - 2

# Its derivative: f'(x) = 2x.
def f_prime(x):
    return 2 * x

# Quick look: f(1) is negative and f(2) is positive, so the root sits between.
print("f(1) =", f(1))
print("f(2) =", f(2))
```

## Step 2 — the Newton–Raphson update rule
From a guess `x`, the rule gives a better guess:

    x_new = x − f(x) / f'(x).

Geometrically: stand at the current guess, slide down the tangent line to where it crosses the x-axis, and take that point as your next guess. We start from a deliberately rough guess of 2 and repeat the rule a few times, printing the sequence.

```python

# Start from a rough guess. The true answer is sqrt(2) = 1.4142135...
x = 2.0

# We will record each guess so we can see the sequence converge.
guesses = [x]

print("step 0 : x =", x)

# Apply the update rule 6 times.
for step in range(1, 7):
    x = x - f(x) / f_prime(x)      # the Newton-Raphson step
    guesses.append(x)
    print("step", step, ": x =", x)
```

## Step 3 — compare with the real √2
Python's `math` library knows √2 precisely. Let us see how close our final guess got. Even after just a few steps the agreement is striking.

```python

import math

true_value = math.sqrt(2)
our_value = guesses[-1]      # the last (best) guess in our list

print("our final guess :", our_value)
print("math.sqrt(2)    :", true_value)
print("difference      :", abs(our_value - true_value))
```

## Step 4 — picture the convergence
If we plot the error |xₙ − √2| at each step on a log scale, it plunges almost straight down. Newton–Raphson roughly doubles the number of correct digits at every step — that is why it is so loved.

```python

import matplotlib.pyplot as plt

# The error at each step: how far that guess is from the true sqrt(2).
errors = []
for guess in guesses:
    errors.append(abs(guess - true_value))

step_numbers = range(len(errors))

plt.figure(figsize=(7, 5))
plt.semilogy(step_numbers, errors, "-o", color="#2E75B6")
plt.xlabel("step number")
plt.ylabel("error |x - sqrt(2)|  (log scale)")
plt.title("Newton-Raphson converges very fast")
plt.show()
```

## What you just did
You implemented Newton–Raphson with a small loop that applies one update rule, `x = x - f(x)/f'(x)`, and you watched the guesses converge to √2. This is your first numerical method: start with a guess, improve it, repeat until it is good enough.
