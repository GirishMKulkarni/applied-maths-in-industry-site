make a much better guess, and repeat. In about five steps we are correct to many
decimal places.

## Step 1 — the idea, in plain words

Suppose your current guess for √2 is a little too big. Then `2 / guess` is a little
too small, because the two must multiply to 2. So the true answer sits somewhere
between `guess` and `2 / guess`. The obvious move is to take the midpoint, the
average of the two, as your next guess. Then repeat.

That single "average your guess with `2 / guess`" step is our whole method. This
same guess-improve-repeat idea is everywhere in industry: it is how a calculator
finds an interest rate that fits a loan, and how a GPS pins your position from
satellite signals.

```python
# One improvement step: turn a guess into a better guess.
def better_guess(guess):
    return (guess + 2 / guess) / 2

# Try it once from a deliberately rough starting point.
print("start with 2.0")
print("one step gives", better_guess(2.0))
```

## Step 2 — repeat the step and watch it home in

Start from a rough guess of 2 and apply the rule a handful of times, printing each
guess. Watch the digits lock into place, one line at a time. We keep every guess in
a list so we can plot them shortly.

```python
guess = 2.0
guesses = [guess]

print("step 0 : guess =", guess)
for step in range(1, 6):
    guess = better_guess(guess)     # improve it
    guesses.append(guess)
    print("step", step, ": guess =", guess)
```

## Step 3 — check against the real answer

Python's `math` library knows √2 precisely. Let us see how close our final guess
got. Even after just a few steps, the agreement is striking.

```python
true_value = math.sqrt(2)
our_value = guesses[-1]      # the last, and best, guess in our list

print("our final guess :", our_value)
print("math.sqrt(2)    :", true_value)
print("difference      :", abs(our_value - true_value))
```

## Step 4 — a picture of the guesses settling down

Plot each guess against its step number. The dots drop from the rough start and
flatten onto the true value (the dashed line) almost immediately. That flattening
is what "it converged" means.

```python
step_numbers = range(len(guesses))

plt.figure(figsize=(7, 5))
plt.plot(step_numbers, guesses, "-o", color="#2E75B6", label="our guess")
plt.axhline(true_value, color="#2E9E5B", linestyle="--", label="true sqrt(2)")
plt.xlabel("step number")
plt.ylabel("guess")
plt.title("The guesses settle onto sqrt(2) in a few steps")
plt.legend()
plt.show()
```

### Stretch (optional) — this is Newton-Raphson

Skip this if you are new to the maths. For readers who want the calculus, the
"average with `2 / guess`" rule is a famous general method in disguise, called
**Newton-Raphson**. To solve `f(x) = 0`, you stand at a guess, slide down the
tangent line to where it crosses the axis, and take that point as the next guess.
The update rule is

    x_new = x - f(x) / f'(x).

For √2 we are solving `f(x) = x² - 2 = 0`, whose derivative is `f'(x) = 2x`.
Substitute those in and simplify, and out drops exactly `(x + 2/x) / 2`, the rule
we used. Below is the general version, which finds the root of *any* function you
hand it. We use it to find a cube root too.

```python
def newton_raphson(f, f_prime, start, steps=8):
    x = start
    for step in range(steps):
        x = x - f(x) / f_prime(x)   # slide down the tangent to a better guess
    return x

# Re-derive sqrt(2): solve x^2 - 2 = 0.
sqrt2 = newton_raphson(lambda x: x * x - 2, lambda x: 2 * x, start=2.0)
print("sqrt(2) by Newton-Raphson:", sqrt2)

# And a cube root: solve x^3 - 27 = 0, whose answer is 3.
cbrt27 = newton_raphson(lambda x: x**3 - 27, lambda x: 3 * x**2, start=4.0)
print("cube root of 27         :", cbrt27)
```

### Stretch (optional) — why it is so fast

For the curious: plot the error at each step on a log scale and it plunges almost
straight down. Newton-Raphson roughly *doubles* the number of correct digits at
every step, which is why numerical libraries love it.

```python
errors = []
for one_guess in guesses:
    errors.append(abs(one_guess - true_value))

plt.figure(figsize=(7, 5))
plt.semilogy(range(len(errors)), errors, "-o", color="#C0392B")
plt.xlabel("step number")
plt.ylabel("error |guess - sqrt(2)|  (log scale)")
plt.title("The error collapses: correct digits roughly double each step")
plt.show()
```
