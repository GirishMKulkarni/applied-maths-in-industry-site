"The right kind" means the right **distribution**, the pattern of how often each value
shows up. Here we meet the five distributions you will use again and again, draw a

right one for your problem is where the real judgement lives.

```python
import numpy as np                 # random numbers and fast maths
import matplotlib.pyplot as plt      # drawing charts

# Set a seed so the random samples are the same every time.
np.random.seed(0)

# How many numbers to draw from each distribution.
sample_size = 5000

print("Ready. We will draw", sample_size, "numbers from each distribution.")
```

## Distribution 1 — Uniform: every value equally likely

A **uniform** distribution gives every value in a range an equal chance. Think of a
lucky draw at a company event, or a UPI cashback where any amount between two limits
is equally likely. It is what you reach for when you know the limits and nothing else.
Here we draw numbers between 0 and 1.

```python
# Draw 5000 numbers, each equally likely to be anywhere between 0 and 1.
uniform_sample = np.random.uniform(0, 1, size=sample_size)

# A histogram counts how many samples fall into each bar (bin).
plt.figure(figsize=(7, 4))
plt.hist(uniform_sample, bins=25, color="#2E75B6", edgecolor="white")
plt.xlabel("value")
plt.ylabel("how often")
plt.title("Uniform: every value between 0 and 1 is equally likely")
plt.show()
```

## Distribution 2 — Normal: the bell curve

A **normal** distribution is the famous bell curve. It shows up whenever a value is
the sum of many small effects: people's heights, measurement noise, a single day's
Nifty return. It has a centre (the mean) and a width (the standard deviation). Here
the centre is 0 and the width is 1.

```python
# Draw 5000 numbers from a bell curve centred at 0 with width 1.
normal_sample = np.random.normal(0, 1, size=sample_size)

plt.figure(figsize=(7, 4))
plt.hist(normal_sample, bins=30, color="#27AE60", edgecolor="white")
plt.xlabel("value")
plt.ylabel("how often")
plt.title("Normal: a bell curve, centred at its mean")
plt.show()
```

## Distribution 3 — Exponential: waiting times

An **exponential** distribution models the *gap of time between random events*: the
minutes until the next customer walks into an SBI branch, or the next UPI payment
hits a merchant. Most gaps are short, but every so often there is a long lull, which
gives it a long tail stretching to the right. We use this one for arrivals and service
times in the queue notebook.

```python
# Draw 5000 waiting times. The "1.0" is the average gap between events.
exponential_sample = np.random.exponential(1.0, size=sample_size)

plt.figure(figsize=(7, 4))
plt.hist(exponential_sample, bins=30, color="#E67E22", edgecolor="white")
plt.xlabel("waiting time")
plt.ylabel("how often")
plt.title("Exponential: time between random events (long tail)")
plt.show()
```

## Distribution 4 — Poisson: counting events in a window

A **Poisson** distribution counts *how many events happen in a fixed window*: how
many customers walk into a bank branch in an hour, how many parcels land at a
Delhivery hub in a minute. It only takes whole-number values, because you cannot have
half a customer. The average count here is 4.

```python
# Draw 5000 counts, each the number of events in a window where the average is 4.
poisson_sample = np.random.poisson(4, size=sample_size)

# Because the values are whole numbers, we line the bars up on the integers.
plt.figure(figsize=(7, 4))
plt.hist(poisson_sample, bins=range(0, 16), color="#8E44AD", edgecolor="white")
plt.xlabel("number of events")
plt.ylabel("how often")
plt.title("Poisson: count of events in a window (whole numbers)")
plt.show()
```

## Distribution 5 — Binomial: successes out of n tries

A **binomial** distribution counts *how many successes happen in n yes/no trials*,
each with the same chance `p`. Think of how many of 10 items at a Flipkart packing
station are damaged, or how many of 10 loan applicants clear a check. Here we use
n = 10 trials with a success chance of p = 0.4.

```python
# Draw 5000 counts of successes, each out of 10 trials with success chance 0.4.
binomial_sample = np.random.binomial(10, 0.4, size=sample_size)

plt.figure(figsize=(7, 4))
plt.hist(binomial_sample, bins=range(0, 12), color="#C0392B", edgecolor="white")
plt.xlabel("number of successes (out of 10)")
plt.ylabel("how often")
plt.title("Binomial: successes in 10 trials, each with chance 0.4")
plt.show()
```

## A quick summary table

One line on what each distribution models. Choosing the right one for your problem is
the real modelling judgement. The computer will happily simulate nonsense if you feed
it the wrong distribution.

```python
print("Uniform     : every value in a range is equally likely (a lucky draw).")
print("Normal      : a sum of many small effects (heights, noise, a day's return).")
print("Exponential : the time between random events (gap between arrivals).")
print("Poisson     : the count of events in a window (customers per hour).")
print("Binomial    : the number of successes in n yes/no trials.")
```

### Stretch (optional) — watch the sample settle onto the theory

Skip this if you are new to code. If you are comfortable, here is the law of large
numbers from notebook 1, wearing a different hat. Each distribution has a true average
we can write down. As we draw a bigger and bigger sample, the sample's own average
should march towards that true value.

The exponential we drew has average equal to its scale, which was 1.0. Let us draw
growing samples and watch the sample mean close in on 1.0.

```python
# Draw one big exponential sample, then look at the average of the first N values
# for several growing values of N.
np.random.seed(0)
big_sample = np.random.exponential(1.0, size=200000)

for how_many in [100, 1000, 10000, 200000]:
    sample_mean = np.mean(big_sample[:how_many])
    print("first", how_many, "values -> sample average:", round(sample_mean, 4))

print()
print("The true average is 1.0. The more values we include, the closer we get.")
print("That is the law of large numbers, the same rule behind the pi estimate.")
```
