## Step 1 — a list of numbers

The simplest thing NumPy gives you is an **array**: a list of numbers you can do
maths on all at once. Here are the sizes, in square feet, of five flats. We hand
the plain list to `np.array` and get back an array.

Every array knows two things about itself. Its **shape** says how big it is (five
numbers in a row, written `(5,)`). Its **dtype** says what kind of number is inside
(here, whole numbers).

```python
# Five flat sizes, in square feet.
flat_sizes = np.array([650, 820, 500, 1100, 940])

print("the sizes :", flat_sizes)
print("shape     :", flat_sizes.shape)   # (5,) means five numbers in a row
print("dtype     :", flat_sizes.dtype)   # the kind of number inside
```

## Step 2 — maths on the whole list at once

Here is the point of the whole session. The business wants to raise every price by
8 percent. We do not walk down the list one flat at a time. We multiply the whole
list by 1.08 in a single stroke, and every price updates together. No loop.

Adding two lists works the same way: NumPy adds them position by position.

```python
# Last year's price for each of the five flats, in lakh.
prices_lakh = np.array([55.0, 72.0, 40.0, 98.0, 85.0])

# Raise every price by 8 percent, all in one line.
new_prices = prices_lakh * 1.08

print("old prices :", prices_lakh)
print("new prices :", new_prices.round(2))

# A small fixed charge added to every flat (say a 2 lakh registration cost).
registration = np.array([2.0, 2.0, 2.0, 2.0, 2.0])
total_cost = new_prices + registration

print("total cost :", total_cost.round(2))
```

## Step 3 — a table of numbers

Stack several lists on top of each other and you get a **table**: rows and columns.
NumPy calls it a 2-D array. We build it from a list of lists, where each inner list
is one row.

Here each row is one flat, and the two columns are its size and its age in years.
The shape of a table is `(number of rows, number of columns)`.

```python
# Each row is one flat: [size in sq ft, age in years].
flats_table = np.array([[650,  8],
                        [820,  3],
                        [500, 15],
                        [1100, 1]])

print("the table:")
print(flats_table)
print()
print("shape:", flats_table.shape, " (4 rows, 2 columns)")
```

## Step 4 — quick ways to make arrays

You will not always type numbers by hand. These helpers make common arrays for
you. You give them a size, and they fill it in. They are handy for setting up a
blank result before filling it, or making an evenly spaced range of values.

```python
# A list of three zeros.
zeros_list = np.zeros(3)
print("zeros   :", zeros_list)

# A 2-by-2 table full of ones.
ones_table = np.ones((2, 2))
print("ones table:")
print(ones_table)

# Evenly spaced numbers: start at 0, stop before 10, step of 2.
counting = np.arange(0, 10, 2)
print("arange  :", counting)

# Ten evenly spaced points from 0 to 100 (useful for plotting later).
spread = np.linspace(0, 100, 10)
print("linspace:", spread.round(1))
```

## Step 5 — picking one number

We reach into a list with square brackets. One thing surprises everyone at first:
**counting starts at 0**. So the first flat is at position 0, not position 1. Say
it out loud once and it sticks.

```python
flat_sizes = np.array([650, 820, 500, 1100, 940])

# The first flat is at position 0.
print("first size :", flat_sizes[0])

# The third flat is at position 2.
print("third size :", flat_sizes[2])

# The very last one can be reached with position -1.
print("last size  :", flat_sizes[-1])
```

## Step 6 — picking a range

A slice `start:stop` gives you several in a row. It **includes** the start position
but **stops before** the stop position, so `1:3` gives positions 1 and 2.

```python
flat_sizes = np.array([650, 820, 500, 1100, 940])

# Positions 1 and 2 (stops before position 3).
print("sizes[1:3] :", flat_sizes[1:3])

# From the start up to (but not including) position 2.
print("sizes[:2]  :", flat_sizes[:2])

# From position 2 to the end.
print("sizes[2:]  :", flat_sizes[2:])
```

## Step 7 — pulling a row, a column, or a block from a table

For a table we give two positions in the brackets: `[rows, columns]`. A colon `:`
on its own means "all of this direction". This is how you grab one flat, or the
whole column of sizes, or a small block.

```python
# Each row is one flat: [size, age, price in lakh].
flats = np.array([[650,  8, 55.0],
                  [820,  3, 72.0],
                  [500, 15, 40.0],
                  [1100, 1, 98.0]])

print("the whole table:")
print(flats)

# One row: flat number 1 (the second flat), all its numbers.
print()
print("one flat  (flats[1, :]) :", flats[1, :])

# One column: all rows, column 0 (every flat's size).
print("all sizes (flats[:, 0]) :", flats[:, 0])

# A small block: the first two flats, their size and age columns.
print()
print("a block   (flats[0:2, 0:2]):")
print(flats[0:2, 0:2])
```

## Step 8 — adding a row to every row (broadcasting)

Sometimes you want to combine a small array with a bigger one. Say you want to add
a fixed adjustment to each column of the table: bump every size by 10, leave age
alone, add 5 lakh to every price. You write the adjustment once as a single row,
and NumPy applies it to every row of the table. This stretching of the smaller
array to fit is called **broadcasting**, and it saves you a loop.

```python
# A small table of three flats: [size, age, price].
small_flats = np.array([[650,  8, 55.0],
                        [820,  3, 72.0],
                        [500, 15, 40.0]])

# One adjustment row: +10 to size, +0 to age, +5 to price.
adjustment = np.array([10, 0, 5.0])
print("adjustment row:", adjustment)

# Broadcasting: the one row is added to EVERY row of the table.
adjusted = small_flats + adjustment
print()
print("after adjusting every flat:")
print(adjusted)
```

### Stretch (optional) — a slice shares memory

Skip this if you are new to code. Here is a genuine surprise that catches everyone
once. When you slice out part of an array, you do not get a fresh copy. You get a
**view** into the same numbers. Change the view and you change the original too. If
you truly want a separate copy, call `.copy()`.

```python
original = np.array([10, 20, 30, 40, 50])

# Take a slice, then change its first entry.
part = original[1:4]
part[0] = 999

print("we only edited the slice, but the original changed too:")
print("original :", original)   # notice the 999 leaked back in

# The safe way: ask for an independent copy first.
fresh = original[1:4].copy()
fresh[0] = -1
print("with .copy(), the original is left alone:", original)
```
