## A first look — what an error message is

Below we run one line that fails, and we print the error Python gives us. Notice the
two parts: the **error type** (here `NameError`) and a short **explanation** after
the colon.

```python
# We will reuse this small helper so every example looks the same.
# It runs a piece of code, and if that code fails, it prints the error neatly
# instead of stopping the whole notebook.
def show_error(label, faulty_function):
    print("Example:", label)
    try:
        faulty_function()
    except Exception as the_error:
        # type(the_error).__name__ is the error's family name, e.g. "NameError".
        print("  error type   :", type(the_error).__name__)
        print("  error message:", the_error)
    print()


# Our first faulty piece of code: we use a name we never created.
def use_a_missing_name():
    print(total_amount)          # we never defined total_amount


show_error("using a name that does not exist", use_a_missing_name)
```

### Reading that bottom-up

- **Bottom line:** `NameError: name 'total_amount' is not defined`. That is the real
  problem.
- **The cause:** we used `total_amount` before ever creating it.
- **The fix:** define `total_amount = ...` first, or check for a typo in the name.

## Error 1 — NameError

A `NameError` means you used a name (a variable or a function) that Python has never
heard of. Almost always a **typo**, or you forgot to define it first. Here we set up
`monthly_salary` but then misspell it.

```python
# We meant to use "monthly_salary" but typed "monthly_salry".
def name_error_example():
    monthly_salary = 45000
    print(monthly_salry)         # typo: missing an 'a'


show_error("NameError", name_error_example)

print("How to fix it: check the spelling, or define the name before you use it.")
```

## Error 2 — KeyError

A `KeyError` means you asked a dictionary (or a pandas column) for a key that is not
there. Usually a **spelling mistake** in the key name. Here a price list has `chai`
and `samosa`, but we ask for `coffee`.

```python
# A dictionary of canteen prices in rupees.
def key_error_example():
    prices = {"chai": 10, "samosa": 15}
    print(prices["coffee"])      # there is no key called "coffee"


show_error("KeyError", key_error_example)

print("How to fix it: print the real keys with list(prices) and match the spelling.")
```

## Error 3 — IndexError

An `IndexError` means you asked for an item past the end of a list. Remember: the
first item is at position `0`, so a list of 3 items has positions `0`, `1`, `2`, and
there is no position `3`.

```python
# This list has 3 items, at positions 0, 1 and 2.
def index_error_example():
    top_stocks = ["TCS", "INFY", "HDFC"]
    print(top_stocks[3])         # position 3 does not exist (the last is 2)


show_error("IndexError", index_error_example)

print("How to fix it: check the length with len(top_stocks); the last position is len - 1.")
```

## Error 4 — ValueError

A `ValueError` means the *type* is right but the *value* makes no sense for what you
asked. A classic case: a form field that should hold a number actually holds text,
and you try to turn it into a number.

```python
# int(...) turns text into a whole number - but only if the text is digits.
def value_error_example():
    age_from_form = int("N/A")   # someone typed "N/A" instead of a number


show_error("ValueError", value_error_example)

print("How to fix it: check the value first; convert only text that really is a number.")
```

## Error 5 — TypeError

A `TypeError` means you combined two things that do not go together, like adding a
number to a piece of text. Python does not know how to add them.

```python
# You cannot glue a number onto text with + .
def type_error_example():
    message = "Total is Rs " + 250   # text + number is not allowed


show_error("TypeError", type_error_example)

print('How to fix it: make both sides text, e.g. "Total is Rs " + str(250).')
```

## A longer, real-looking traceback, read bottom-up

When your code calls another function which then fails, Python prints a longer
traceback with several lines. It looks scary, but the rule never changes: **start at
the bottom**, then read *upward* to find the line in *your* code that started it.

Here one function calls a helper, and the helper fails because of a typo in a
dictionary key: it asks for `close_price` but the data has `clse_price` (a real
mistake when reading an NSE price file).

```python
# An outer function that calls an inner helper.
def load_price(one_day):
    return one_day["close_price"]      # the typo is here: the key is "clse_price"


def run_report():
    one_day = {"clse_price": 100}      # note: "clse_price", missing the 'o'
    price = load_price(one_day)        # this call will fail inside load_price
    print(price)


# We run it and print the FULL traceback as text, so you can read it bottom-up.
import traceback

try:
    run_report()
except Exception:
    print(traceback.format_exc())

print("Bottom line says KeyError: 'close_price'.")
print("Read upward: the failing line is one_day['close_price'] inside load_price,")
print("called from run_report. The data only has 'clse_price' - a typo.")
```

### Stretch (optional) — raise your own clear error

Skip this if today is your first week of code. Once you are comfortable reading
errors, you can also *make* them on purpose, with a helpful message, so a bad input
fails loudly and clearly instead of causing a confusing crash later. `raise` does
this. Notice how much friendlier the message is than a random crash.

```python
# A function that refuses a negative price, with a message that explains why.
def record_sale(price):
    if price < 0:
        raise ValueError("price cannot be negative, got " + str(price))
    print("Recorded a sale of Rs", price)


# The good case works.
record_sale(250)

# The bad case fails with OUR clear message. We catch it so the notebook continues.
try:
    record_sale(-40)
except ValueError as the_error:
    print("Refused, as intended. Message:", the_error)
```

## What you just learned

- An error message is information, not failure.
- **Read it bottom-up:** the last line is the *kind* of error and a short message;
  the lines above show *where* it happened.
- The five you will meet most: **NameError** (unknown name), **KeyError** (missing
  key), **IndexError** (past the end), **ValueError** (wrong value), **TypeError**
  (wrong kind of thing).
- Once you know the family, the fix is usually obvious. Stay calm and read the clue.
