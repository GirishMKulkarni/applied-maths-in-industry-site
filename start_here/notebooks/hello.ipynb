## Step 2 — a tiny sum
The computer is a very fast, very literal calculator. Let us add up a day's orders.

```python
orders = [250, 90, 400, 120, 700]   # five orders in rupees
total = 0
for one_order in orders:
    total = total + one_order
print("Total sales today: Rs", total)
```

## Step 3 — your first chart
One short line draws a picture of those orders. Seeing your data is a habit we keep all course.

```python
import matplotlib.pyplot as plt   # the drawing library

plt.figure(figsize=(6, 4))
plt.bar(range(1, 6), orders, color="#2E75B6")
plt.xlabel("order number")
plt.ylabel("amount (Rs)")
plt.title("Today's orders")
plt.show()
```
